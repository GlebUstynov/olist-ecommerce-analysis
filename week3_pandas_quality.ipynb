### Выводы по качеству данных

1. Ключи `orders.order_id`, `customers.customer_id` и `products.product_id` уникальны. В `order_items` уникальна комбинация `(order_id, order_item_id)`. Полных дубликатов ключей не обнаружено. При этом `customer_unique_id` повторяется 3 345 раз, но это не ошибка: один реальный клиент может оформить несколько заказов.

2. В таблицах `customers` и `order_items` пропусков нет. В `orders` обнаружено 4 908 пропущенных ячеек: 160 в `order_approved_at`, 1 783 в `order_delivered_carrier_date` и 2 965 в `order_delivered_customer_date`.

3. В `products` обнаружено 2 448 пропущенных ячеек. По 610 пропусков содержат `product_category_name`, `product_name_lenght`, `product_description_lenght` и `product_photos_qty`. В столбцах веса, длины, высоты и ширины товара обнаружено по два пропуска.

4. Большинство отсутствующих дат доставки являются ожидаемыми для заказов со статусами `canceled`, `unavailable`, `shipped`, `invoiced`, `processing` и `created`, поскольку такие заказы не были полностью доставлены. Отсутствие даты доставки у заказа со статусом `delivered` следует считать потенциальной проблемой качества данных.

5. Обнаружено 775 заказов без позиций: 603 со статусом `unavailable`, 164 — `canceled`, 5 — `created`, 2 — `invoiced` и 1 — `shipped`. Большинство таких случаев связано с отменёнными или недоступными заказами, которые не дошли до нормального завершения. Оставшиеся случаи могут указывать на неполноту исходных данных.

6. Все заказы успешно сопоставились с таблицей клиентов, а все позиции заказов — с таблицей товаров. Проверки `validate` подтвердили ожидаемые связи `one-to-many` и `many-to-one`.

7. Результаты Pandas полностью совпали с PostgreSQL: 112 650 строк, 98 666 уникальных заказов, 95 420 реальных клиентов, общая стоимость товаров 13 591 643,70 и стоимость доставки 2 251 909,54.


```python
def table_quality(df, table_name, key_columns):
    return {
        'table_name' : table_name,
        'row_count' : len(df),
        'column_count' : df.shape[1],
        'duplicate_key_count' : df.duplicated(subset=key_columns).sum(),
        'missing_cells' : df.isna().sum().sum()
    }
```

```python
table_quality(orders, 'orders', ['order_id'])
```

```python
quality_report_function = pd.DataFrame([
    table_quality(orders, 'orders', ['order_id']),
    table_quality(customers, 'customers', ['customer_id']),
    table_quality(order_items, 'order_items', ['order_id', 'order_item_id']),
    table_quality(products, 'products', ['product_id'])
])

quality_report_function
```
