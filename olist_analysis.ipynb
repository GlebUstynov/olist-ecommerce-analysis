##  Data Cleaning

```python
orders["order_purchase_timestamp"] = pd.to_datetime(orders["order_purchase_timestamp"])
orders["order_approved_at"]=pd.to_datetime(orders["order_approved_at"])
orders["order_delivered_carrier_date"]=pd.to_datetime(orders["order_delivered_carrier_date"])
orders["order_delivered_customer_date"]=pd.to_datetime(orders["order_delivered_customer_date"])
orders["order_estimated_delivery_date"]=pd.to_datetime(orders["order_estimated_delivery_date"])
```

```python
for name, df in {
    "customers": customers,
    "order_items": order_items,
    "products": products
}.items():

    print(f"===== {name.upper()} OVERVIEW =====")

    print("\nHEAD:")
    print(df.head())

    print("\nSHAPE:")
    print(df.shape)

    print("\nCOLUMNS:")
    print(df.columns)

    print("\nDTYPES:")
    print(df.dtypes)

    print("\nMISSING VALUES:")
    print(df.isna().sum())

    print("\nDUPLICATES:")
    print(df.duplicated().sum())

    print("\n")
```

```python
orders_customers = pd.merge(orders, customers, on="customer_id", how="left")
```

```python
print("orders:", orders.shape)
print("orders_customers:", orders_customers.shape)
```

```python
orders_customers.head()
```

## Data Preparation and Merging

```python
sales_data = pd.merge(
    orders_customers,
    order_items,
    on="order_id",
    how="left"
)
```

```python
print("orders_customers:", orders_customers.shape)
print("sales_data:", sales_data.shape)
```

```python
final_data = pd.merge(
    sales_data,
    products,
    on="product_id",
    how="left"
)
```

```python
print("sales_data:", sales_data.shape)
print("final_data:", final_data.shape)
```

```python
final_data.head()
```

## Business analysis

### 5.1 Total Revenue

```python
total_revenue = final_data["price"].sum()
total_revenue
```

### 5.2 Revenue by State

```python
top_states = (
    final_data
    .groupby("customer_state")["price"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_states
```

```python
top_states_pct = (top_states / final_data["price"].sum()) * 100

print(top_states_pct.round(2))
```

### 5.3 Average Order Value

```python
avg_order_value = (
    final_data
    .groupby("order_id")["price"]
    .sum()
    .mean()
)

avg_order_value
```

### 5.4 Order Status Distribution

```python
order_status_distribution = orders["order_status"].value_counts()
order_status_distribution
```

```python

order_status_pct = orders["order_status"].value_counts(normalize=True) * 100

print(order_status_pct.round(2))
```

### 5.5 Top Customers

```python
top_customers = (
    final_data
    .groupby("customer_unique_id")["price"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_customers
```

### 5.6 Monthly Revenue

```python
final_data["month"] = final_data["order_purchase_timestamp"].dt.to_period("M")
```

```python
final_data[["order_purchase_timestamp", "month"]].head()
```

```python
monthly_revenue = (
    final_data
    .groupby("month")["price"]
    .sum()
    .sort_index()
)

monthly_revenue
```

### 5.7 Top Product Categories

```python
top_categories = (
    final_data
    .groupby("product_category_name")["price"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_categories
```

### 5.8 Category Performance

```python
category_performance = (
    final_data
    .groupby("product_category_name")
    .agg(
        total_revenue=("price", "sum"),
        item_count=("order_item_id", "count")
    )
    .sort_values("total_revenue", ascending=False)
    .head(10)
)

category_performance
```

##  Key Insights

- The majority of orders were successfully delivered (97.02%).

- Revenue is concentrated in several states. SP has the highest share of total revenue (38.28%), followed by RJ (13.42%) and MG (11.66%). Together, these three states account for more than half of total revenue.

- The average order value is around $136.68.

- There is a noticeable difference in spending between customers, with some customers spending significantly more than others.

- Some product categories generate significantly more revenue than others. The top categories are `beleza_saude` (1,258,681.34), `relogios_presentes` (1,205,005.68), and `cama_mesa_banho` (1,036,988.68).

- Monthly revenue changes noticeably over time. The peak was in November 2017, while October 2016 had the lowest meaningful revenue, excluding October 2018.
